```python
# 帮我写一个检查df是否有数值相同的两列
def check_duplicate_columns(df, return_details=False):
    """
    检查DataFrame中是否存在数值相同的列
    
    参数:
        df (pd.DataFrame): 输入的DataFrame
        return_details (bool): 是否返回详细信息，默认False返回简要结果
    
    返回:
        如果 return_details=False:
            dict: {'has_duplicates': bool, 'duplicate_pairs': list}
        如果 return_details=True:
            dict: {'has_duplicates': bool, 'duplicate_pairs': list, 'similarity_matrix': pd.DataFrame}
    
    示例:
        >>> df = pd.DataFrame({'A': [1,2,3], 'B': [1,2,3], 'C': [4,5,6]})
        >>> check_duplicate_columns(df)
        {'has_duplicates': True, 'duplicate_pairs': [('A', 'B')]}
    """
    
    n_cols = len(df.columns)
    duplicate_pairs = []
    similarity_matrix = pd.DataFrame(1.0, index=df.columns, columns=df.columns)
    
    for i in range(n_cols):
        for j in range(i + 1, n_cols):
            col_i = df.iloc[:, i]
            col_j = df.iloc[:, j]
            
            # 检查数值是否完全相同
            if col_i.equals(col_j):
                duplicate_pairs.append((df.columns[i], df.columns[j]))
                similarity = 1.0
            else:
                # 计算余弦相似度（可选）
                similarity = float(np.corrcoef(col_i, col_j)[0, 1]) if len(col_i) > 1 else 0
            
            similarity_matrix.iloc[i, j] = similarity
            similarity_matrix.iloc[j, i] = similarity
    
    result = {
        'has_duplicates': len(duplicate_pairs) > 0,
        'duplicate_pairs': duplicate_pairs,
        'n_duplicates': len(duplicate_pairs)
    }
    
    if return_details:
        result['similarity_matrix'] = similarity_matrix
    
    return result


def print_duplicate_report(df):
    """
    打印重复列的详细报告
    
    参数:
        df (pd.DataFrame): 输入的DataFrame
    """
    result = check_duplicate_columns(df, return_details=True)
    
    print("\n" + "="*60)
    print("列重复性检查报告")
    print("="*60)
    print(f"数据形状: {df.shape}")
    print(f"总列数: {len(df.columns)}")
    print(f"重复列对数: {result['n_duplicates']}")
    
    if result['has_duplicates']:
        print("\n【发现重复列对】")
        for pair in result['duplicate_pairs']:
            print(f"  ✓ '{pair[0]}' 与 '{pair[1]}' 数值相同")
            # 打印前5个值作为示例
            print(f"    示例值: {df[pair[0]].head(3).tolist()}")
    else:
        print("\n✓ 未发现重复列")
    
    # 显示相似度矩阵
    print("\n【列相似度矩阵（余弦相关系数）】")
    print(result['similarity_matrix'].round(3).to_string())
    print("="*60 + "\n")
```

```python
result_a = check_duplicate_columns(ddi_a, return_details=True)
result_v = check_duplicate_columns(ddi_v, return_details=True)
result_shi = check_duplicate_columns(shi, return_details=True)
result_dehi = check_duplicate_columns(dehi, return_details=True)
result_cluster = check_duplicate_columns(cluster, return_details=True)

print_duplicate_report(ddi_a)
print_duplicate_report(ddi_v)
print_duplicate_report(shi)
print_duplicate_report(dehi)
print_duplicate_report(cluster)
```
